### MacLaurin series for $e^x$

```python
x = np.linspace(0, 10, 100)
y = np.exp(x)
```

```python
plt.figure()

plt.plot(x, y, 'k--')

plt.show()
```

```python
# Your turn: pick a function and slide the number of terms. SymPy builds the series;
# the plot shows each term, their sum, and the real function (like the plots above).
%matplotlib inline
import numpy as np
from matplotlib import pyplot as plt
import sympy as sp
from ipywidgets import interact

X = sp.symbols('x')
functions = {'e^x': sp.exp(X), 'sin(x)': sp.sin(X), 'cos(x)': sp.cos(X)}
xs = np.linspace(-5, 5, 400)

def values(expression):
    return 0 * xs + sp.lambdify(X, expression)(xs)   # numbers to plot; the 0*xs turns a constant into a flat line

def plot_series(function='sin(x)', n_terms=1):
    f = functions[function]
    # the Maclaurin series as a plain polynomial, then its terms in order of increasing power
    polynomial = f.series(X, 0, 40).removeO()
    terms = sorted(sp.Add.make_args(polynomial), key=lambda term: sp.degree(term, X))[:n_terms]
    plt.figure()
    for term in terms:
        plt.plot(xs, values(term), label=f'${sp.latex(term)}$')
    plt.plot(xs, values(sum(terms)), 'purple', linewidth=3, label='sum of the terms')
    plt.plot(xs, values(f), 'k--', label=function)
    plt.ylim(-10, 10)
    plt.legend(loc='upper left', frameon=False)
    plt.show()

interact(plot_series, function=list(functions), n_terms=(1, 10))
```
