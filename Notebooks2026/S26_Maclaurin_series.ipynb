```python
%matplotlib inline
import numpy as np
from matplotlib import pyplot as plt
```

### MacLaurin series for $e^x$

```python
x = np.linspace(0, 10, 100)
y = np.exp(x)
```

```python
plt.figure()

plt.plot(x, y, 'k--')

plt.show()
```

```python
y0 = 1
y1 = x
y2 = (1./2.) * x**2
y3 = (1./6.) * x**3

f = y0 + y1 + y2 + y3
```

```python
plt.figure()

plt.axhline(y=y0, color='red', label='f0')

plt.legend(loc='upper left', frameon=False)

plt.ylim(-1,10)

plt.show()
```

```python
plt.figure()

plt.axhline(y=y0, color='red', label='f0')
plt.plot(x, y1, 'orange', label='f1')

plt.legend(loc='upper left', frameon=False)


plt.show()
```

```python
plt.figure()

plt.axhline(y=y0, color='red', label='f0')
plt.plot(x, y1, 'orange', label='f1')
plt.plot(x,y2, 'green', label='f2')

plt.legend(loc='upper left', frameon=False)


plt.show()
```

```python
plt.figure()

plt.axhline(y=y0, color='red', label='f0')
plt.plot(x, y1, 'orange', label='f1')
plt.plot(x,y2, 'green', label='f2')
plt.plot(x,y3, 'blue', label='f3')

plt.legend(loc='upper left', frameon=False)


plt.show()
```

```python
plt.figure()

plt.axhline(y=y0, color='red', label='f0')
plt.plot(x, y1, 'orange', label='f1')
plt.plot(x,y2, 'green', label='f2')
plt.plot(x,y3, 'blue', label='f3')
plt.plot(x,f, 'purple', label='f sum')

plt.legend(loc='upper left', frameon=False)


plt.show()
```

```python
plt.figure()

plt.axhline(y=y0, color='red', label='f0')
plt.plot(x, y1, 'orange', label='f1')
plt.plot(x,y2, 'green', label='f2')
plt.plot(x,y3, 'blue', label='f3')
plt.plot(x,f, 'purple', label='f sum')
plt.plot(x,y, 'k--', label='e^x')

plt.legend(loc='upper left', frameon=False)


plt.ylim(-5,200)

plt.show()
```

```python
plt.figure()

plt.axhline(y=y0, color='red', label='f0')
plt.plot(x, y1, 'orange', label='f1')
plt.plot(x,y2, 'green', label='f2')
plt.plot(x,y3, 'blue', label='f3')
plt.plot(x,f, 'purple', label='f sum')
plt.plot(x,y, 'k--', label='e^x')

plt.legend(loc='upper left', frameon=False)

plt.ylim(-5, 25)
plt.xlim(0, 3)

plt.show()
```

### MacLaurin series for $sin(x)$

```python
y = np.sin(x)
```

```python
plt.figure()

plt.plot(x, y, 'k--')

plt.show()
```

```python
y0 = 0
y1 = x
y2 = 0
y3 = (-1./6.)*x**3
y4 = 0
y5 = (1./120.)*x**5
```

```python
plt.figure()

plt.plot(x, y1, 'red', label='f1')
plt.plot(x,y3, 'orange', label='f3')
plt.plot(x,y5, 'green', label='f5')
plt.plot(x,y1+y3+y5, 'purple', label='f sum')

plt.legend(loc='upper left', frameon=False)

plt.show()
```

```python
plt.figure()

plt.plot(x, y1, 'red', label='f1')
plt.plot(x,y3, 'orange', label='f3')
plt.plot(x,y5, 'green', label='f5')
plt.plot(x,y1+y3+y5, 'purple', label='f sum')
plt.plot(x,y, 'k--', label='sin(x)')

plt.legend(loc='lower right', frameon=False)

plt.ylim(-3, 3)
#plt.xlim(0, 3)

plt.show()
```
