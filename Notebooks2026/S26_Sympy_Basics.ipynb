We can create some function y which is a function of x: 

```python
y = Function('y')(x)
```

```python
y
```

We can specify that the function is $y(x) = e^x$:

```python
y = exp(x)
```

```python
y
```

```python
print(y)
```

What does this do?

```python
y.diff(x)
```

I can set this result equal to a new variable called result:

```python
result = y.diff(x)
```

```python
result
```

Let's double check that the derivative equals the original function (which it should for an exponential function!)

```python
y.diff(x).equals(y) 
```

## Math manipulation
If we write a mathemtical expression, we can use SymPy to expand, factor, and simplify!

```python
math_expression = x+2*y
math_expression
```

```python
x * math_expression
```

```python
expand( x * math_expression)
```

```python
simplify(expand( x * math_expression))
```

```python
factor(expand( x * math_expression)) 
```

## basic calculus

We can also do calculus with SymPy!

```python
expr = sin(x) * exp(x)
expr
```

```python
diff(expr)
```

```python
integrate( expr, x)
```

```python
expr = exp(x)*sin(x) + exp(x) * cos(x)
expr
```

```python
integrate(expr, x)
```

```python
limit(sin(x)/x, x, 0) 
```
