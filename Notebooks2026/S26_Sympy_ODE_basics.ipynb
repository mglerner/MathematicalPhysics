## SymPy homogeneous ODE solution

#### specify variables and function

```python
# specify what we will use as symbolic variables and functions
from sympy.abc import x, y
y = Function('y')
y
```

**Important detail**

> In the SymPy Basics notebook we wrote `y = Function('y')(x)`. Here we write `y = Function('y')`.
>
> With `y = Function('y')(x)`, every later `y` implicitly means $y(x)$. That is convenient, but it is not required. The alternative is to define `y = Function('y')` and then write `y(x)` explicitly wherever we need it.
>
> When solving differential equations, the second form is usually easier to keep track of: define `y = Function('y')`, and write `y(x)` whenever we mean "$y$ as a function of $x$". It also lets us write `y(0)` for an initial condition.

#### Let's create a homogeneous ODE.

We can just provide an expression for the left-hand side of the equation. When we do this, it is presumed that the right-hand side equals 0.

I would like to write the following ODE: $9y(x) + \frac{d^2}{dx^2}y(x)$

How can I do this with SymPy?

```python
y(x)
```

```python
y(x).diff(x)
```

```python
y(x).diff(x,1)
```

```python
y(x).diff(x,2)
```

```python
ODE_LHS = y(x).diff(x,2) + 9*y(x) 
ODE_LHS
```

Now I want to solve my differential equation! This means I want an equation for $y(x)$. I can use SymPy for this!

```python
# We need to specify the ODE and the function we want to solve for. 
result = dsolve(ODE_LHS, y(x)) 
result
```

Let's check that the solution works! "check ode sol" plugs in our result to the left-hand side of the ODE and compares to the right-hand side (which should be 0).

```python
checkodesol(ODE_LHS, result)
# returns True or False and "value after substitution"
```

## SymPy inhomogeneous ODE solution

### first, modify command to specify an equation instead of an expression for the ODE
For an inhomogeneous equation, we need to specify both the LHS and the RHS. We therefore now specify an equation instead of an expression. To do this, we use  `Eq(LHS, RHS)`.

```python
# define the same homogeneous ODE as before, but this time as an equation using `Eq()`
ODE_Eq = Eq(ODE_LHS, 0)
ODE_Eq
```

```python
result = dsolve(ODE_Eq, y(x))
result
```

```python
# Now let's try an inhomogeneous ODE: 
ODE_Eq = Eq(ODE_LHS, 27)
ODE_Eq
```

```python
result = dsolve(ODE_Eq, y(x))
result
```

## SymPy ODE solution with initial and/or boundary conditions

### How to set boundary conditions: some examples
- $f^{\prime}(1) = 2 \rightarrow$ `ics={f(x).diff(x).subs(x, 1): 2}`
- $f(0) =1$ and $f^\prime(2) = 3 \rightarrow$ `ics={f(0): 1, f(2): 3}`
   

```python
ODE_Eq = Eq(ODE_LHS, 0)
result = dsolve(ODE_Eq, y(x))
result
```

Let's set some conditions such that at $t=0$, $y=1$, and $y' = 1$, and then solve for our arbitrary constants:

```python
ODE_Eq = Eq(ODE_LHS, 0)
result = dsolve(ODE_LHS, y(x), ics={y(0): 1, y(x).diff(x).subs(x,0):1})
result
```

```python
ODE_Eq = Eq(y(x).diff(x,1) , exp(x) / sqrt(k*y(x)))
result = dsolve(ODE_Eq, y(x) )
result
```

```python
# let's check this result
# evaluate the LHS of this ODE
result[0].rhs.diff(x)
```

```python
left.equals(right)
```

```python
ODE_Eq = Eq(y(x).diff(x) + y(x), exp(x)* (y(x))**2)
ODE_Eq
```

```python
problem_result = dsolve(ODE_Eq, y(x) )
problem_result
```

```python
# verify solution
# first evaluate LHS of ODE
left = problem_result.rhs + problem_result.rhs.diff(x) 
left
```

```python
#now evaluate RHS of ODE

right = (problem_result.rhs)**2 * exp(x)
right
```

```python
left.equals(right)
```
