#### Did this do what you expect? Note that the n that you give this function is not the number of terms it wants, but the highest order (the highest exponent in your polynomial). This means that for some series, you will need to figure out which power of n will be the last term you save. Sympy is quick, so you can just plug in different n values until you get the number of terms you want, if it's not immediately clear.


```python
maclaurin_series = f_x.series(x, 0, 7)
maclaurin_series
```

```python
#If you want to get rid of that note about higher orders, you can add .removeO() after your series function. Try it below:
maclaurin_series = f_x.series(x, 0, 7).removeO()
maclaurin_series
```

### Series Manipulation

```python
# Differentiating a series to find another series
f_x = sin(x)
sin_series = (f_x.series(x, 0, 6).removeO())
cos_series = diff(f_x.series(x, 0, 6).removeO(),x)
```

```python
sin_series
```

```python
cos_series
```

```python
# Integrating a series to find another series
f_x = cos(x)
cos_series = (f_x.series(x, 0, 6).removeO())
sin_series = integrate(f_x.series(x, 0, 6).removeO(),x)

```

```python
cos_series
```

```python
sin_series
```
