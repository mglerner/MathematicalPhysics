## 2. Demo: one block of code, several parameter values

Here's the trick we need, shown on a *different* (first-order) equation:

$$\frac{dy}{dt} + k\,y = 0, \qquad y(0) = 2,$$

for three different decay rates $k = 0.5$, $k = 1$, and $k = 2$.

First, a short function that takes a value of $k$ and hands back SymPy's solution for that
value. This is just the `dsolve(..., ics=...)` you've seen before, wrapped up in a function so
we can call it again and again.

```python
t = sp.symbols('t')
y = Function('y')

def solve_demo(k_value):
    ODE_LHS = y(t).diff(t) + k_value * y(t)
    return dsolve(ODE_LHS, y(t), ics={y(0): 2})

solve_demo(2)
```

Now the loop. Here's a line-by-line explanation. You should read through this and understand
it ... and then feel free to copy and modify it for your own purposes.

- `t_smooth = np.linspace(0, 5, 400)` is 400 evenly spaced times from 0 to 5, like in WHW04.
- `[0.5, 1, 2]` is a Python **list**: values in square brackets, separated by commas.
- `for k_value in [0.5, 1, 2]:` runs the indented lines three times, with `k_value` equal to
  0.5 the first time, 1 the second time, and 2 the third time. (In WHW04 we looped over
  `range(n_steps)`, which is the numbers 0, 1, 2, ...; here we loop over whatever list we like.)
- Inside the loop we solve, `lambdify` the right-hand side (exactly as in WHW04), and plot.
- `label=f'k = {k_value}'` is an **f-string**: the `f` in front means "replace `{k_value}` with
  its actual value", so the three curves get the labels `k = 0.5`, `k = 1`, and `k = 2`.
- `plt.axhline(0, color='gray')` draws a horizontal line at zero, which makes it easier to see
  when a curve gets there.
- The `plt.xlabel` ... `plt.show()` lines are *not* indented, so they run once, after the loop
  is done.

```python
t_smooth = np.linspace(0, 5, 400)

for k_value in [0.5, 1, 2]:
    solution = solve_demo(k_value)
    solution_numpy = sp.lambdify(t, solution.rhs, 'numpy')
    plt.plot(t_smooth, solution_numpy(t_smooth), label=f'k = {k_value}')

plt.axhline(0, color='gray')
plt.xlabel('t (s)')
plt.ylabel('y')
plt.legend()
plt.show()
```

## 3. Your turn: the damped oscillator

Now do the same thing for

$$x'' + b\,x' + 4x = 0, \qquad x(0) = 1, \qquad x'(0) = -2 .$$

Two lines of `solve_oscillator` need to change:

1. `ODE_LHS` should be the left-hand side of *our* ODE. You need a second derivative; look
   back at how `S26_Sympy_ODE_basics.ipynb` wrote $y''$.
2. The `ics` need **two** initial conditions, one on $x(0)$ and one on $x'(0)$. The section
   "SymPy ODE solution with initial and/or boundary conditions" of `S26_Sympy_ODE_basics.ipynb`
   shows how to write a condition on a derivative.

As handed to you, the cell runs, but it solves the demo's first-order equation. That's on purpose:
change it!

```python
x = Function('x')

def solve_oscillator(b_value):
    # Change the line below.
    ODE_LHS = x(t).diff(t) + b_value * x(t)
    # Change the line below.
    return dsolve(ODE_LHS, x(t), ics={x(0): 1})
```

Let's look at the three solutions. (`display` is how you ask Jupyter to show something nicely
from *inside* a loop; at the bottom of a cell it happens automatically.) If you want, add a
`checkodesol` check here too.

```python
for b_value in [1, 4, 6]:
    print('b =', b_value)
    display(solve_oscillator(b_value))
```

## 4. Now let's plot it and see if we were right.

This is the demo's loop, pointed at your function. It plots all three solutions for
$0 \le t \le 5$ on one set of axes.

```python
t_smooth = np.linspace(0, 5, 400)

for b_value in [1, 4, 6]:
    solution = solve_oscillator(b_value)
    solution_numpy = sp.lambdify(t, solution.rhs, 'numpy')
    plt.plot(t_smooth, solution_numpy(t_smooth), label=f'b = {b_value}')

plt.axhline(0, color='gray')
plt.xlabel('t (s)')
plt.ylabel('x')
plt.legend()
plt.show()
```
