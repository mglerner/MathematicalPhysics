## 2. A grid of points (recap)

As in the WHW10 starter: `np.meshgrid` turns a 1D array of $x$ values and a 1D
array of $y$ values into two 2D arrays `X` and `Y`, one entry per grid point. With
`indexing='xy'`, rows go with $y$ and columns go with $x$. We'll put an arrow at

```python
x_vals = np.linspace(-2, 2, 21)
y_vals = np.linspace(-2, 2, 21)
X, Y = np.meshgrid(x_vals, y_vals, indexing='xy')
print(X.shape)
```

## 3. Drawing a vector field: `quiver` and `streamplot`

First, the field itself as a short named function that hands back the two
components. Written this way, the **same** function works on NumPy arrays
(for plots) and on SymPy symbols (for derivatives), so we only type it once.

```python
def G(x, y):
    return x + y**2, x**2 * y

U, V = G(X, Y)
print(U.shape, V.shape)
```

Now two pictures of $\mathbf{G}$ side by side:

- `plt.subplots(1, 2, figsize=(11, 5))` makes one figure with a row of two
  panels and hands back the figure and an array of the two panels' **axes**.
  We draw on a panel with `axes[0].something(...)` instead of `plt.something(...)`.
- `quiver(X, Y, U, V)` draws an arrow at each point `(X, Y)` pointing along
  `(U, V)`. Matplotlib shrinks all the arrows by the same factor so they fit;
  longer arrow still means stronger field.
- `streamplot(X, Y, U, V)` draws curves that are everywhere tangent to the
  field: the paths a speck of dust would follow. It wants the
  `indexing='xy'` grid, which is what we made.
- `set_aspect('equal')` keeps circles looking like circles.

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 5))
axes[0].quiver(X, Y, U, V)
axes[0].set_title('G with quiver')
axes[1].streamplot(X, Y, U, V)
axes[1].set_title('G with streamplot')
for ax in axes:
    ax.set_xlabel('x')
    ax.set_ylabel('y')
    ax.set_aspect('equal')
plt.show()
```

## 4. Divergence and curl in SymPy

In two dimensions, for $\mathbf{F} = (F_x, F_y)$,

$$\nabla \cdot \mathbf{F} = \frac{\partial F_x}{\partial x} + \frac{\partial F_y}{\partial y},
\qquad (\nabla \times \mathbf{F})_z = \frac{\partial F_y}{\partial x} - \frac{\partial F_x}{\partial y}.$$

Same field function as before, now handed SymPy symbols instead of arrays.
`diff(expression, x)` is the partial derivative with respect to `x`.

```python
x, y = sp.symbols('x y')

def divergence_and_curl(field):
    Fx, Fy = field(x, y)
    divergence = simplify(diff(Fx, x) + diff(Fy, y))
    curl_z = simplify(diff(Fy, x) - diff(Fx, y))
    return divergence, curl_z

divergence_and_curl(G)
```

## 6. Flux through the edges of a square, as a Riemann sum (demo)

The outward flux through a closed curve is $\oint \mathbf{F} \cdot \hat{\mathbf{n}}\, ds$:
add up (outward component of the field) $\times$ (length of each little piece)
all the way around. For a square, the outward normal $\hat{\mathbf{n}}$ is simple
on each edge:

| edge | $\hat{\mathbf{n}}$ | $\mathbf{F}\cdot\hat{\mathbf{n}}$ |
|---|---|---|
| right | $(1, 0)$ | $F_x$ |
| left | $(-1, 0)$ | $-F_x$ |
| top | $(0, 1)$ | $F_y$ |
| bottom | $(0, -1)$ | $-F_y$ |

Demo: the flux of $\mathbf{G}$ out of the **unit square** $0 \le x, y \le 1$. Chop
each edge into `n_edge` pieces of length `ds` and sample at the middle of each
piece, just like the midpoint sums in the WHW10 starter.

- `s_mid` is the array of midpoints along an edge, running from 0 to 1.
- `np.ones(n_edge)` is an array of `n_edge` ones. On the right edge every point
  has $x = 1$, so we hand the field `1.0 * ones` for $x$ and `s_mid` for $y$.
  (Why not just `1.0`? For a field like $(x, y)$, $F_x$ would then come back as the
  single number 1.0 instead of one value per piece, and `np.sum` would add up only
  one piece.)
- Each edge's flux is `np.sum(` outward component `) * ds`.

```python
def flux_demo(n_edge):
    ds = 1.0 / n_edge
    s_mid = (np.arange(n_edge) + 0.5) * ds
    ones = np.ones(n_edge)
    # right edge, x = 1, outward normal (1, 0)
    Fx, Fy = G(1.0 * ones, s_mid)
    flux_right = np.sum(Fx) * ds
    # left edge, x = 0, outward normal (-1, 0)
    Fx, Fy = G(0.0 * ones, s_mid)
    flux_left = np.sum(-Fx) * ds
    # top edge, y = 1, outward normal (0, 1)
    Fx, Fy = G(s_mid, 1.0 * ones)
    flux_top = np.sum(Fy) * ds
    # bottom edge, y = 0, outward normal (0, -1)
    Fx, Fy = G(s_mid, 0.0 * ones)
    flux_bottom = np.sum(-Fy) * ds
    return flux_right + flux_left + flux_top + flux_bottom
```

The divergence theorem says the outward flux equals the integral of the
divergence over the inside of the square. SymPy's double integral, as in the
WHW10 starter (one tuple per variable):

```python
divergence_G, curl_G = divergence_and_curl(G)
div_theorem_demo = integrate(divergence_G, (x, 0, 1), (y, 0, 1))
print('divergence theorem:', div_theorem_demo, '=', float(div_theorem_demo))
for n_edge in [4, 10, 100]:
    print('n_edge =', n_edge, '  Riemann-sum flux =', flux_demo(n_edge))
```
